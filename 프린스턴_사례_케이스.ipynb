### 1.3 표준화 점수 (z-score)

한가지 기준을 더 고려해봅시다.  
정규분포시간에 어디에 쓰는지 모르겠는데 배운 그거, `표준화 점수`를 사용해봅시다.

```python
white_df["z_score"] = (
    white_df.groupby("Judge")["Rating"]
           .transform(lambda s: (s - s.mean()) / s.std(ddof=1))
)

red_df["z_score"] = (
    red_df.groupby("Judge")["Rating"]
           .transform(lambda s: (s - s.mean()) / s.std(ddof=1))
)
```

```python
red_df.head(10)
```

####z-score 결과

```python
white_df.groupby(['Wine'])[['Rating','Rank', 'z_score']].mean().sort_values('z_score', ascending=False)
```

```python
red_df.groupby(['Wine'])[['Rating','Rank', 'z_score']].mean().sort_values('z_score', ascending=False)
```

## 원점수의 특징
1)	원점수는 간단하고 설명하기 쉽다
2)	하지만 주요 단점은, 점수를 매우 다양하게 주는 심사위원들이 1점에서 20점까지의 점수 범위 중 좁은 일부만을 사용하는 심사위원들보다 최종 순위에 더 큰 영향을 미칩니다
3)	이를 확인하기 위해, 모든 와인에 10점을 주는 심사위원을 생각해보십시오. 그 심사위원은 대회의 결과에 아무런 영향을 미치지 않습니다. 반대 극단으로, 한 와인에 20점을 주고 나머지 와인들에 1점을 준 심사위원은 자신이 가장 선호하는 와인을 거의 확실하게 높은 순위에 올릴 수 있겠죠

우리의 레드 와인 데이터에서, 와인 E는 원점수 기준으로는 4위를 차지했지만(표준화 점수나 순위를 사용하면 7위), 이는 상당 부분 이 와인이 올리비에 게르고(Olivier Gerguad)가 가장 선호하는 와인이었기 때문이죠. 그는 20점 척도의 비교적 넓은 범위를 사용한 심사위원이었고, 따라서 원점수 기반 경쟁에서 더 큰 영향력을 가지고 있습니다.

## 원점수의 대안

만약 모든 심사위원이 동등한 영향력을 갖는 것이 중요하다면(즉, 그들이 20점 척도를 어떻게 사용했는지와 무관하게 동등한 가중치를 부여받아야 한다면), 순위나 표준화 점수를 사용할 수 있습니다.

1)	순위는 이해하기 쉽지만 일부 정보가 손실된다는 단점이 있습니다.
2)	예를 들어, 한 심사위원으로부터 12, 11, 2점을 받은 세 와인은 각각 1, 2, 3위의 순위를 얻게 됩니다. 하지만 12, 11, 10점을 받은 와인들도 마찬가지로 1, 2, 3위의 순위를 받습니다. 분명히 첫 번째 경우에 심사위원은 세 번째 와인이 두 번째 경우의 세 번째 와인보다 훨씬 나쁘다고 말하고 있지만, 순위는 이 정보를 무시합니다.


## 표준화 점수

1)	서로 다른 척도를 갖고 있지만 그 정보를 보존하고 싶으면 어떻게 해야할까요?? 이럴 때는 표준화 점수를 활용합니다.
2)	표준화 점수는 정보를 보존하지만, 설명하기가 더 어렵다는 단점이 있습니다.



## 원점수가 아닐 때의 단점
표준화 점수(Scaled scores)와 순위(Ranks)를 사용하는 평가 방식에는 또 다른 단점이 있습니다.

- 바로 '무관한 대안의 독립성(independence of irrelevant alternatives)'이라는 공리를 만족시키지 못한다는 점입니다.

- 예를 들어, 제가 사과와 바나나 중에서 사과를 더 좋아한다고 결정했다고 해봅시다. 이때 누군가 와서 "오렌지도 있는데, 같이 고려해보세요"라고 해도, 제 마음속의 '사과 > 바나나'라는 순위는 변하지 않아야 한다는 뜻입니다. 오렌지라는 '무관한 대안'이 추가되거나 사라지는 것이 사과와 바나나 사이의 선호도에 영향을 주어서는 안 된다는 거죠.


## 무관한 대안의 독립성 예시

image.png

## 상황 1: 와인 A, B, C가 모두 경쟁할 때

먼저 세 와인이 모두 참가했을 때의 최종 순위를 보겠습니다. 각 심사위원의 순위를 모두 더해서 총합이 가장 낮은 와인이 우승합니다.
 - 와인 A 총점: 1 + 1 + 2 + 1 + 2 + 2 + 2 = 11점
- 와인 B 총점: 3 + 3 + 1 + 3 + 1 + 1 + 1 = 13점
- 와인 C 총점: 2 + 2 + 3 + 2 + 3 + 3 + 3 = 18점

이때의 순위는 A > B > C 입니다.


## 상황 2: 꼴찌였던 와인 C를 제외하고 경쟁할 때

이제 여기서 '무관한 대안', 즉 어차피 꼴찌였던 와인 C를 빼고 A와 B만으로 순위를 다시 매겨보겠습니다. 심사위원들의 A와 B에 대한 선호도 자체는 변하지 않았지만, C가 사라졌으니 둘만 놓고 1등과 2등을 가리는 겁니다.
- 심사위원 1: A를 B보다 선호 → A(1등), B(2등)
- 심사위원 2: A를 B보다 선호 → A(1등), B(2등)
- 심사위원 3: B를 A보다 선호 → B(1등), A(2등)
- 심사위원 4: A를 B보다 선호 → A(1등), B(2등)
- 심사위원 5: B를 A보다 선호 → B(1등), A(2등)
- 심사위원 6: B를 A보다 선호 → B(1등), A(2등)
- 심사위원 7: B를 A보다 선호 → B(1등), A(2등)

이제 이 새로운 순위들의 총합을 계산해 보겠습니다.
- 와인 A 총점: 1 + 1 + 2 + 1 + 2 + 2 + 2 = 11점
- 와인 B 총점: 2 + 2 + 1 + 2 + 1 + 1 + 1 = 10점
놀랍게도 이번에는 총점이 더 낮은 와인 B가 우승합니다. 순위가
B > A 로 뒤바뀐 것입니다.


어차피 꼴찌였던 C를 뺐을 뿐인데, A와 B의 우열이 뒤바뀌었습니다. 이것이 바로 '무관한 대안의 독립성' 원칙이 깨진 사례입니다.
이러한 현상은 순위 기반 평가 방식의 허점을 보여줍니다. 참가자(여기서는 와인) 중 일부를 추가하거나 제외하는 것만으로 전체 결과가 바뀔 수 있기 때문입니다.

z점수도 마찬가지입니다.
Z-점수 = (개별 와인의 원점수 - 해당 심사위원의 평균 점수) / 해당 심사위원의 표준편차
따라서 와인 i의 점수가 빠지면 각 심사위원의 평균 점수 및 표준편차가 모두 변하게 됨 -> 순위 변동 가능성 생김

이러한 순위 역전 현상은  총 원점수(total raw scores)를 사용할 때는 일어나지 않습니다.

# 4번째 질문

## 최고의 심사위원은 누구인가?

#### = 가장 평균에 근사한 사람

이 질문이야말로 다양한 의견이 있을 것 같습니다.

제 한가지 의견은 다수의 의견,  
즉 집단지성의 값에 가장 근접한 예측을 한 사람이 객관적인 눈을 가진 최고의 심사위원이라고 생각합니다.

이건 어떻게 계산하면 될까요? 이 역시 분산이나 표준편차를 이용하면 될 것 같군요.

개별 심사위원의 점수가 전체 심사위원의 점수로부터 얼마나 떨어져있는지 확인해봅시다.

```python
white_df["mean"] = white_df.groupby("Wine")["z_score"].transform("mean")
white_df["dev"] = white_df["z_score"] - white_df["mean"]
white_df["std_z"] = white_df["dev"]**2
```

```python
red_df["mean"] = red_df.groupby("Wine")["z_score"].transform("mean")
red_df["dev"] = red_df["z_score"] - red_df["mean"]
red_df["std_z"] = red_df["dev"]**2
```

```python
white_df.head(10)
```

```python
red_df.head(20)
```

```python
red_df.groupby(['Judge'])['std_z'].sum().sort_values(ascending=True)
```

```python
white_df.groupby(['Judge'])['std_z'].sum().sort_values(ascending=True)
```
